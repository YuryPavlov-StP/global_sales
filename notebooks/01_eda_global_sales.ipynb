```python
df[['Region', 'Salesperson', 'Revenue','Profit']][df['Revenue'] == 172962]
```

```python
# df[['Region', 'Salesperson', 'Revenue']][df['Revenue'] == 172962] - это цепочка из двух последовательных индексаций
# Первый [...] возвращает DataFrame, второй [...] применяется уже к нему, а не к исходному df. 
# Визуально выглядит как «два df», но на самом деле это один объект, к которому дважды применили [].
# Правильно вот так: 2 варианта
df.loc[df['Revenue'] == 172962, ['Region', 'Salesperson', 'Revenue']]
```

```python
# или через маску
mask = df['Revenue'] == 172962
df.loc[mask, ['Region', 'Salesperson', 'Revenue', 'Profit']]
```

```python
df.pivot_table(
    index = "Region", aggfunc={"Units_Sold": "sum", "Profit": ("sum","mean")}
).sort_values(by=("Units_Sold", "sum"), ascending = False)
```

```python
# Усложнённый инсайт
# - Влияние скидок на выручку/прибыль
# - Обнаружение аномалий
```

```python
find_anomaly_list = ['Revenue',	'Profit',	'Units_Sold',	'Customer_Satisfaction',	'Marketing_Spend',	'Discount_Percentage']
for col in find_anomaly_list:
    plt.figure(figsize=(10,4))
    plt.title(col)
    plt.boxplot(df[col], vert=False)
    plt.show()
```

```python
plt.scatter(df['Discount_Percentage'], df['Revenue'])
```
